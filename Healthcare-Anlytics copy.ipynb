## Variables:
The dataset includes a wide range of health indicators, lifestyle factors, and demographic information. Key variables include:

### **Target Variable:**
  * **Diabetes Status (`Diabetes_012`)**:
    * 0 = No diabetes
    * 1 = Prediabetes
    * 2 = Diabetes

### **Health Indicators:**
  * **High Blood Pressure (`HighBP`)**: Binary (0 = No, 1 = Yes)
  * **High Cholesterol (`HighChol`)**: Binary (0 = No, 1 = Yes)
  * **Cholesterol Check (`CholCheck`)**: Binary (0 = No, 1 = Yes)
  * **Body Mass Index (`BMI`)**: Continuous (range: 12–98)
  * **Smoker (`Smoker`)**: Binary (0 = No, 1 = Yes)
  * **Stroke (`Stroke`)**: Binary (0 = No, 1 = Yes)
  * **Heart Disease or Attack (`HeartDiseaseorAttack`)**: Binary (0 = No, 1 = Yes)

### **Lifestyle Factors:**
  * **Physical Activity (`PhysActivity`)**: Binary (0 = No, 1 = Yes)
  * **Fruit Consumption (`Fruits`)**: Binary (0 = No, 1 = Yes)
  * **Vegetable Consumption (`Veggies`)**: Binary (0 = No, 1 = Yes)
  * **Heavy Alcohol Consumption (`HvyAlcoholConsump`)**: Binary (0 = No, 1 = Yes)

### **Healthcare Access:**
  * **Any Healthcare (`AnyHealthcare`)**: Binary (0 = No, 1 = Yes)
  * **No Doctor Due to Cost (`NoDocbcCost`)**: Binary (0 = No, 1 = Yes)

### **Demographics:**
  * **Sex (`Sex`)**: Binary (0 = Female, 1 = Male)
  * **Age (`Age`)**: Categorical (13 age groups, e.g., 1 = 18–24, 13 = 80+)
  * **Education (`Education`)**: Categorical (6 levels, e.g., 1 = Never attended school, 6 = College graduate)
  * **Income (`Income`)**: Categorical (8 levels, e.g., 1 = <$10,000, 8 = ≥$75,000)

### **Mental and Physical Health:**
  * **General Health (`GenHlth`)**: Categorical (1 = Excellent, 5 = Poor)
  * **Mental Health (`MentHlth`)**: Continuous (days of poor mental health in the past 30 days, range: 0–30)
  * **Physical Health (`PhysHlth`)**: Continuous (days of poor physical health in the past 30 days, range: 0–30)
  * **Difficulty Walking (`DiffWalk`)**: Binary (0 = No, 1 = Yes)

## Limitations:
  * **Self-Reported Data**: The dataset relies on self-reported health behaviors and conditions, which may introduce bias (e.g., underreporting or overreporting).
  * **Categorical Variables**: Some variables (e.g., `Income`, `Education`) are categorical and may require encoding for analysis.
  * **Missing Values**: While the dataset appears clean, some variables (e.g., `BMI`, `MentHlth`, `PhysHlth`) may have missing or outlier values that need to be addressed.
  * **Cross-Sectional Data**: The dataset represents a single point in time (2015), so it cannot be used to establish causality or track changes over time.
  * **Generalizability**: While the dataset is large and diverse, it may not fully represent all populations (e.g., specific ethnic groups or regions).

```python
fig = px.histogram(
    df,
    x='Diabetes_012',
    color='Sex',
    barmode='group',
    title='Distribution of Diabetes_012 by Gender',
    category_orders={"Diabetes_012": [0, 1, 2]},
    color_discrete_map={0: 'pink', 1: 'blue'}  # Map colors for Sex=0 (Female) and Sex=1 (Male)
)
fig.show()
```

```python



# Define mappings for gender colors and labels
sex_colors = {0: 'pink', 1: 'blue'}
sex_labels = {0: 'Female', 1: 'Male'}

# Create a 4x4 grid of subplots with titles for ages 1 to 13
fig = make_subplots(rows=4, cols=4, subplot_titles=[f"Age: {age}" for age in range(1, 14)])

# Loop through ages 1 to 13
for i, age in enumerate(range(1, 14)):
    row = i // 4 + 1
    col = i % 4 + 1
    
    # Filter data for the current age and group by Diabetes_012 and Sex
    group = df[df['Age'] == age].groupby(['Diabetes_012', 'Sex']).size().unstack(fill_value=0)
    
    # Add a bar for each sex category
    for sex in [0, 1]:
        fig.add_trace(
            go.Bar(
                x=group.index.astype(str),  # Convert diabetes categories to string labels
                y=group[sex] if sex in group.columns else [0] * len(group.index),
                name=f"Sex: {sex_labels[sex]}",
                marker_color=sex_colors[sex],
                showlegend=(i == 0)  # Only show legend in the first subplot
            ),
            row=row,
            col=col
        )

# Update layout and display the figure
fig.update_layout(
    barmode='group',
    title_text="Diabetes Distribution by Age and Sex",
    height=1200,
    width=1500
)
fig.for_each_yaxis(lambda yaxis: yaxis.update(type="log"))
fig.show()
```

```python



# Define the columns for subplots
columns = ["GenHlth", "MentHlth", "Age", "Education"]
sex_colors = {0: 'pink', 1: 'blue'}
sex_labels = {0: 'Female', 1: 'Male'}

# Create a 2x2 subplot figure
fig = sp(rows=2, cols=2, subplot_titles=columns)

# Add grouped bar charts to subplots
for i, column in enumerate(columns, start=1):
    row, col = (i - 1) // 2 + 1, (i - 1) % 2 + 1  # Convert index to subplot grid

    # Group data by column and Sex
    grouped_data = df.groupby([column, 'Sex']).size().unstack(fill_value=0)

    # Plot each sex category
    for sex, color in sex_colors.items():
        fig.add_trace(
            go.Bar(
                x=grouped_data.index.astype(str),  # Convert x values to string for display
                y=grouped_data[sex] if sex in grouped_data.columns else [0] * len(grouped_data),
                name=f"Sex: {sex_labels[sex]}",
                text=grouped_data[sex],
                textposition='auto',
                marker=dict(color=color)
            ),
            
            row=row, col=col
        )

# Update layout for better appearance
fig.update_layout(
    title_text="Health Data Subplots - Frequency Distribution by Sex",
    showlegend=True,
    height=1000,
    width=1500
)

# Show the plot
fig.for_each_yaxis(lambda yaxis: yaxis.update(type="log"))
fig.show()
```

```python
# Data cleaning
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
#remove duplicates
df = df.drop_duplicates()
```

```python
# Function to detect outliers using IQR and return a boolean mask
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    # Return a boolean mask where True indicates non-outliers
    return (df[column] >= lower_bound) & (df[column] <= upper_bound)

# Example: Detect and drop outliers in the 'BMI' column
# Create a boolean mask for non-outliers
non_outliers_mask = detect_outliers('BMI')

# Filter the DataFrame to keep only non-outliers
df_cleaned = df[non_outliers_mask]

# Print the cleaned DataFrame
print("DataFrame after dropping outliers in 'BMI':\n", df_cleaned)
```

# Visualization after cleaning

```python
def detect_outliers(column, threshold=1.5):
    Q1 = df[column].quantile(0.25)  # First quartile (25th percentile)
    Q3 = df[column].quantile(0.75)  # Third quartile (75th percentile)
    IQR = Q3 - Q1  # Interquartile range
    lower_bound = Q1 - threshold * IQR  # Lower bound for outliers
    upper_bound = Q3 + threshold * IQR  # Upper bound for outliers
    # Return a boolean mask where True indicates non-outliers
    return (df[column] >= lower_bound) & (df[column] <= upper_bound)

# Example: Detect and drop outliers in the 'BMI' column with a custom threshold
threshold = 1.5  # Adjust this value (e.g., 1.5, 3, etc.)
non_outliers_mask = detect_outliers('BMI', threshold=threshold)

# Filter the DataFrame to keep only non-outliers
df_cleaned1 = df[non_outliers_mask]

# Print the cleaned DataFrame
print("DataFrame after dropping outliers in 'BMI':\n", df_cleaned1)
```
